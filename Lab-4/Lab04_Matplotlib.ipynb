## 1. Introduction to Matplotlib
The `pyplot` module is used to draw line plots, scatter plots, bar charts, histograms, pie charts, and subplots.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

## 2. Line Plot
A line plot visualizes trends or changes in numerical data by connecting data points with lines.

```python
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

plt.plot(x, y)
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

## 3. Scatter Plot
A scatter plot shows the relationship between two numerical variables, with each observation drawn as a point.

```python
plt.scatter([1, 2, 3], [5, 7, 6])

plt.title("Scatter Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.show()
```

## 4. Bar Chart
A bar chart compares values across different categories.

```python
categories = ["A", "B", "C"]
values = [12, 18, 10]

plt.bar(categories, values)

plt.title("Bar Chart")
plt.xlabel("Categories")
plt.ylabel("Values")

plt.show()
```

## 5. Histogram
A histogram displays the frequency distribution of numerical values by grouping observations into intervals called *bins*.

```python
data = [1, 2, 2, 3, 3, 3, 4]

plt.hist(data)

plt.title("Histogram")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()
```

## 6. Pie Chart
A pie chart represents proportions of categories. The `autopct` parameter displays each percentage.

```python
values = [40, 35, 25]
labels = ["A", "B", "C"]

plt.pie(
    values,
    labels=labels,
    autopct="%1.1f%%"
)

plt.title("Pie Chart")

plt.show()
```

## 7. Subplots
`plt.subplots(1, 2)` creates one row with two plots; `ax[0]` and `ax[1]` refer to each plot, and `plt.tight_layout()` adjusts spacing.

```python
fig, ax = plt.subplots(1, 2)

# Histogram
ax[0].hist([1, 2, 2, 3, 3, 3, 4])
ax[0].set_title("Histogram")

# Pie chart
ax[1].pie(
    [40, 35, 25],
    labels=["A", "B", "C"],
    autopct="%1.1f%%"
)
ax[1].set_title("Pie Chart")

plt.tight_layout()
plt.show()
```

## 8. Dataset: Titanic Passengers

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the Titanic dataset
df = pd.read_csv("train.csv")

# Display the first five rows
print(df.head())
```

## 9. List of Problems

### Problem 1 – Load and Inspect the Dataset
Load the dataset, display the first five rows, inspect it, and identify numerical and categorical attributes suitable for visualization.

```python
print(df.head())
print("Shape:", df.shape)
df.info()

numerical_cols = df.select_dtypes(include="number").columns.tolist()
categorical_cols = df.select_dtypes(exclude="number").columns.tolist()

print("Numerical attributes:", numerical_cols)
print("Categorical attributes:", categorical_cols)
```

**Interpretation:** Numerical attributes such as `Age`, `Fare`, `SibSp`, and `Parch` suit scatter plots and histograms. Categorical or discrete attributes such as `Sex`, `Pclass`, `Survived`, and `Embarked` suit bar and pie charts. `PassengerId`, `Name`, `Ticket`, and `Cabin` are identifiers or text and are not useful for these plots. `Age` has missing values, which Matplotlib skips automatically.

### Problem 2 – Line Plot: Average Fare by Passenger Class

```python
avg_fare_by_class = df.groupby("Pclass")["Fare"].mean()

plt.plot(avg_fare_by_class.index, avg_fare_by_class.values, marker="o")
plt.title("Average Fare by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Average Fare")
plt.xticks([1, 2, 3])
plt.show()
```

**Interpretation:** The average fare falls steeply from first class (about 84) to second (about 21) and third (about 14), showing that ticket price is closely tied to class.

### Problem 3 – Scatter Plot: Age vs Fare

```python
plt.scatter(df["Age"], df["Fare"], alpha=0.6)
plt.title("Passenger Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```

**Interpretation:** There is no strong linear relationship between age and fare. Most passengers paid low fares across all ages, and a few high-fare outliers (above 200) appear among middle-aged passengers.

### Problem 4 – Bar Chart: Male vs Female Passengers

```python
gender_counts = df["Sex"].value_counts()

plt.bar(gender_counts.index, gender_counts.values)
plt.title("Number of Passengers by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.show()
```

**Interpretation:** Male passengers (577) outnumber female passengers (314) by almost two to one.

### Problem 5 – Histogram: Distribution of Passenger Ages

```python
plt.hist(df["Age"].dropna(), bins=20, edgecolor="black")
plt.title("Distribution of Passenger Ages")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

**Interpretation:** Ages are concentrated between about 20 and 40, with a smaller peak among young children and a long tail of older passengers.

### Problem 6 – Pie Chart: Survived vs Did Not Survive

```python
survival_counts = df["Survived"].value_counts().sort_index()

plt.pie(
    survival_counts.values,
    labels=["Did not survive", "Survived"],
    autopct="%1.1f%%"
)
plt.title("Proportion of Survivors")
plt.show()
```

**Interpretation:** About 61.6% of passengers did not survive, while about 38.4% survived.

### Problem 7 – Two Subplots: Age Histogram and Passengers per Class

```python
fig, ax = plt.subplots(1, 2, figsize=(11, 4))

# Histogram of ages
ax[0].hist(df["Age"].dropna(), bins=20, edgecolor="black")
ax[0].set_title("Distribution of Passenger Ages")
ax[0].set_xlabel("Age")
ax[0].set_ylabel("Frequency")

# Bar chart of class counts
class_counts = df["Pclass"].value_counts().sort_index()
ax[1].bar(class_counts.index.astype(str), class_counts.values, edgecolor="black")
ax[1].set_title("Passengers per Class")
ax[1].set_xlabel("Passenger Class")
ax[1].set_ylabel("Number of Passengers")

plt.tight_layout()
plt.show()
```

**Interpretation:** Most passengers are young adults, and third class holds the largest group (491), far more than first (216) or second (184).

### Problem 8 – Three Subplots: Gender, Fare, and Survival

```python
fig, ax = plt.subplots(1, 3, figsize=(11, 4))

# 1. Bar chart: passengers by gender
gender_counts = df["Sex"].value_counts()
ax[0].bar(gender_counts.index, gender_counts.values, edgecolor = "black")
ax[0].set_title("Passengers by Gender")
ax[0].set_xlabel("Gender")
ax[0].set_ylabel("Number of Passengers")

# 2. Histogram: fare distribution
ax[1].hist(df["Fare"], bins=30, edgecolor="black")
ax[1].set_title("Distribution of Fares")
ax[1].set_xlabel("Fare")
ax[1].set_ylabel("Frequency")

# 3. Pie chart: survival proportion
survival_counts = df["Survived"].value_counts().sort_index()
ax[2].pie(
    survival_counts.values,
    labels=["Did not survive", "Survived"],
    autopct="%1.1f%%"
)
ax[2].set_title("Survival Proportion")

plt.tight_layout()
plt.show()
```

**Interpretation:** The fare distribution is strongly right-skewed, since most passengers paid low fares and a few paid very high ones. Together with the gender and survival panels, the figure gives a compact overview of who was aboard and how many survived.
