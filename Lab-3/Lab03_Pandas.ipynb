## 3. Selecting Data

```python
# One column
print(df["Name"].head())

# Several columns
print(df[["Name", "Age", "Sex", "Survived"]].head())

# Rows by label (loc) and by position (iloc)
print(df.loc[0:4, ["Name", "Age"]])
print(df.iloc[0:5, 0:4])
```

## Problem 2 – Data Filtering

**1. First-class passengers**

```python
first_class = df[df["Pclass"] == 1]
first_class.head()
```

**2. Passengers younger than 18**

```python
young = df[df["Age"] < 18]
young.head()
```

**3. Male passengers who survived**

```python
male_survived = df[(df["Sex"] == "male") & (df["Survived"] == 1)]
male_survived.head()
```

**4. Fare greater than 100**

```python
high_fare = df[df["Fare"] > 100]
high_fare.head()
```

## Problem 3 – Sorting and Aggregation

**1. Top 10 highest fares**

```python
top_fares = df.sort_values(by="Fare", ascending=False)
top_fares[["Name", "Pclass", "Fare"]].head(10)
```

**2. Average age**

```python
print("Average age:", df["Age"].mean())
```

**3. Male and female count**

```python
print(df["Sex"].value_counts())
```

**4. Average fare by class**

```python
print(df.groupby("Pclass")["Fare"].mean())
```

**5. Survival rate by class**

```python
print(df.groupby("Pclass")["Survived"].mean())
```

### Findings
